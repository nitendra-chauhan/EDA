```python
## Calculating outliers using IQR method

for col in num_col:
    Q1 = df[col].quantile(0.25) 
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    print(f"{col}:")
    print(f" Lower Bound = {lower_bound}")
    print(f" Upper Bound = {upper_bound}")
    print(f" Number of Outliers = {outliers.shape[0]}\n")
```

```python
# Alternate
for col in numeric_data:
    Q1 = numeric_data[col].quantile(0.25)
    Q3 = numeric_data[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = numeric_data[(numeric_data[col] < lower_bound) | (numeric_data[col] > upper_bound)]
    print(f"{col}:")
    print(f" Lower Bound = {lower_bound}")
    print(f" Upper Bound = {upper_bound}")
    print(f" Number of Outliers = {outliers.shape[0]}\n")
```

```python
org_skewness=df[["order_number", "qty_ordered", "PRICEEACH", "SALES", "DAYS_SINCE_LASTORDER", "MSRP"]].skew()
org_skewness
```

Based on the above skewness results, "SALES" coln is highly skewed.

```python
# Log Transformation
df['log_SALES'] = np.log(df['SALES'] + 1) 

# Square Root Transformation
df['sqrt_SALES'] = np.sqrt(df['SALES'])

# Cube Root Transformation
df['cbrt_SALES'] = np.cbrt(df['SALES'])

# Box-Cox Transformation
from scipy.stats import boxcox
df['box_cox_SALES'], _ = boxcox(df['SALES'] + 1)

# Significance of df['boxcox_SALES']:
#It means create a new column named boxcox_SALES inside the DataFrame df or replace that column if it already exists.
```

```python
transformed_skewness = df[['log_SALES', 'sqrt_SALES', 'cbrt_SALES', 'boxcox_SALES']].skew()   ### Imp line hai bss yhi waali
print("Transformed Skewness:\n", transformed_skewness)
#print(transformed_skewness)

# Visualize the transformed distributions
plt.figure(figsize=(18, 12))

# Original Distribution
plt.subplot(3, 2, 1)
sns.histplot(df['SALES'], kde=True)
plt.title('Original SALES')

# Log Transformation
plt.subplot(3, 2, 2)
sns.histplot(df['log_SALES'], kde=True)
plt.title('Log Transformed SALES')

# Square Root Transformation
plt.subplot(3, 2, 3)
sns.histplot(df['sqrt_SALES'], kde=True)
plt.title('Square Root Transformed SALES')

# Cube Root Transformation
plt.subplot(3, 2, 4)
sns.histplot(df['cbrt_SALES'], kde=True)
plt.title('Cube Root Transformed SALES')
```

Box-cox transformation techniques is the best to rectify the skewness problem.
